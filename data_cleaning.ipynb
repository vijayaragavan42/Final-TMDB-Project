```python
placeholders = [
    "unknown",
    "n/a",
    "na",
    "n.a.",
    "none",
    "null",
    "nan",
    "?",
    "not available",
    "not known"
]  #Define the placeholder values

for col in df5.columns:
    values = df5[col].astype("string").str.strip()

    mask = values.str.casefold().isin(placeholders).fillna(False) #Identify placeholder values

    print(f"\nColumn: {col}")
    print("Placeholder count:", mask.sum())
    print(df5.loc[mask, col].value_counts())
```
