```python
import pandas as pd 
import numpy as np
from pathlib import Path

print("pandas version:",pd.__version__)
print("Libraries imported successfully") #import pandas


DATA_DIR=Path(r"D:\VScode\TMDB_csv_cleaned_file.csv")  #fill the file path


Caste=pd.read_csv(DATA_DIR/"caste_cleaned.csv")
Crew=pd.read_csv(DATA_DIR/"crew_cleaned.csv")
Genres=pd.read_csv(DATA_DIR/"genres_cleaned.csv")
Movie_genres=pd.read_csv(DATA_DIR/"Movie_genres_cleaned.csv")
Movie_keywords=pd.read_csv(DATA_DIR/"movie_keywords_cleaned.csv")
Movies=pd.read_csv(DATA_DIR/"movies_cleaned.csv")
```

```python
invalid_financials = Movies[
    Movies["budget"].isna()
    | Movies["revenue"].isna()
    | (Movies["budget"] <= 0)
    | (Movies["revenue"] <= 0)
]

invalid_financials[
    ["title", "budget", "revenue"]
]
```

```python
valid_movies = Movies[
    Movies["budget"].notna()
    & Movies["revenue"].notna()
    & (Movies["budget"] > 0)
    & (Movies["revenue"] > 0)
].copy()

valid_movies["profit"] = (
    valid_movies["revenue"] - valid_movies["budget"]
)

valid_movies["roi_percent"] = (
    valid_movies["profit"] / valid_movies["budget"]
) * 100

valid_movies[
    ["title", "budget", "revenue", "profit", "roi_percent"]
]
```

```python
print(valid_movies.columns.tolist())
```

```python
valid_movies=valid_movies.copy()

valid_movies["profit"]=(
    valid_movies["revenue"]-
    valid_movies["budget"]
)

valid_movies["roi_percent"]=(
    (valid_movies["revenue"] -
     valid_movies["budget"])
     /valid_movies["budget"]
)*100
```

```python
top_15_roi = (
    valid_movies[valid_movies["budget"] >= 1_000_000]
    .sort_values("roi_percent", ascending=False)
    .head(15)
)

top_15_roi[
    ["title", "budget", "revenue", "profit", "roi_percent"]
]
```

```python
Movies["release_date"] = pd.to_datetime(
    Movies["release_date"],
    errors="coerce"
)

Movies["release_year"] = Movies["release_date"].dt.year

Movies_per_year = (
    Movies.groupby("release_year")
    .size()
    .reset_index(name="movie_count")
    .sort_values("release_year")
)

display(Movies_per_year)
```

```python
highest_year = Movies_per_year.loc[
    Movies_per_year["movie_count"].idxmax()
]

print("Year with the most movies:", highest_year["release_year"])
print("Number of movies:", highest_year["movie_count"])
```

```python
runtime = pd.to_numeric(Movies["runtime"], errors="coerce")

Q1 = runtime.quantile(0.25)
Q3 = runtime.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

runtime_outliers = Movies[
    (runtime < lower_bound) | (runtime > upper_bound)
]

runtime_outliers[["title", "runtime"]]
```

```python
primary_genres = Genres.drop_duplicates(
    subset="genre_id",
    keep="first"
)

movie_genre_data = Movies.merge(
    primary_genres,
    left_on="movie_id",
    right_on="genre_id",
    how="inner"
)

genre_popularity = (
    movie_genre_data.groupby("genre_name")["popularity"]
    .mean()
    .sort_values(ascending=False)
)

display(genre_popularity)
```

```python
actor_analysis = Caste.merge(
    Movies[["movie_id", "vote_average"]],
    left_on="movie_id",
    right_on="movie_id",
    how="inner"
)

top_10_actors = (
    actor_analysis.groupby("actor_name")
    .agg(
        movie_count=("movie_id", "nunique"),
        average_rating=("vote_average", "mean")
    )
    .sort_values("movie_count", ascending=False)
    .head(10)
)
display(top_10_actors)
```

```python
duplicate_titles = Movies[
    Movies.duplicated(subset=["title"], keep=False)
].sort_values(["title", "release_date"])

duplicate_titles[
    ["title", "release_date"]
]
```

```python
popularity_votes = Movies[
    ["popularity", "vote_count"]
].dropna()

correlation = popularity_votes[
    "popularity"
].corr(popularity_votes["vote_count"])

print("Popularity and vote count correlation:", correlation)
```

```python
popularity_threshold = Movies["popularity"].quantile(0.90)
vote_count_threshold = Movies["vote_count"].quantile(0.25)

unusual_Movies = Movies[
    (
        (Movies["popularity"] >= popularity_threshold)
        & (Movies["vote_count"] <= vote_count_threshold)
    )
    |
    (
        (Movies["popularity"] <= Movies["popularity"].quantile(0.25))
        & (Movies["vote_count"] >= Movies["vote_count"].quantile(0.90))
    )
]

display(
    unusual_Movies[
        ["title", "popularity", "vote_count"]
    ]
)
```
